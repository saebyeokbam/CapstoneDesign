```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
```

```python
temp_df = pd.read_csv("./contents/questions.csv")
```

```python
correct_answer_df = temp_df[['question_id', 'correct_answer']]
```

```python
correct_answer_df
```

```python
columns = correct_answer_df['question_id']
columns
```

```python
res_df = pd.DataFrame(np.full([784309, 13169], False))
res_df.rename(columns=columns,inplace=True)
res_df
```

```python
all_files = os.listdir('./KT1')

for i, fn in enumerate(all_files):
    df = pd.read_csv('./KT1/' + fn)
    for j in df['question_id']:
        correct_answer = correct_answer_df[(correct_answer_df['question_id']==j)]['correct_answer'].values[0]
        user_answer = df[(df['question_id']==j)]['user_answer'].values[0]
        if correct_answer==user_answer:
            res_df[j][i] = True
    if i==10000:
        break
    print(i)
```

```python
res_df_sub = res_df.iloc[:10000]
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

frequent_itemsets = apriori(res_df_sub, min_support=0.05, use_colnames=True)
frequent_itemsets
```

```python
item_list = []
support_list = []

for idx,i in enumerate(frequent_itemsets['itemsets']):
    if len(i) == 2:
        item_list.append(i)
        support_list.append(frequent_itemsets['support'][idx])
```

```python
new_itemsets = pd.DataFrame({'items' : item_list, 'support': support_list})
new_itemsets
```

```python
association_rules(frequent_itemsets, metric="lift", min_threshold=1).head(100)
```

```python
import networkx as nx

G = nx.Graph()
ar = new_itemsets['items']
G.add_edges_from(ar)

pr = nx.pagerank(G)
nsize = np.array([v for v in pr.values()])
nsize = 2000 * (nsize - min(nsize)) / (max(nsize) - min(nsize))

pos = nx.random_layout(G)

plt.figure(figsize=(16, 12))
plt.axis('off')
nx.draw_networkx(G, pos=pos, node_color=list(pr.values()), node_size=nsize, alpha=0.7, edge_color='.5')
plt.show()
```
